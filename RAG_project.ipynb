```python
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

print("Loading model... (first time takes ~30 seconds)")
model = SentenceTransformer('all-MiniLM-L6-v2')
print("✓ Model loaded!")
print(f"\nModel Info:")
print(f"  - Name: all-MiniLM-L6-v2")
print(f"  - Embedding dimension: 384")
print(f"  - Type: Sentence transformer (BERT-based)")
```

```python
# Example sentence
sentence = "Transformers use multi-head attention to process text"

# Create embedding
embedding = model.encode(sentence)

print(f"Sentence: {sentence}")
print(f"\nEmbedding shape (vector dimensions): {embedding.shape}")
print(f"Embedding dtype (data type): {embedding.dtype}")
print(f"\n=== ACTUAL EMBEDDING NUMBERS (First 20 of 384) ===")
print(embedding[:20])
print(f"\n=== ACTUAL EMBEDDING NUMBERS (Last 10 of 384) ===")
print(embedding[-10:])
print(f"\n=== STATISTICS ===")
print(f"Min value: {embedding.min():.4f}")
print(f"Max value: {embedding.max():.4f}")
print(f"Mean value: {embedding.mean():.4f}")
print(f"Std deviation: {embedding.std():.4f}")
```

```python
# Define test sentences
sentences = [
    "Transformers use multi-head attention to process text",
    "Attention mechanisms allow different parts of a sentence to attend to each other",
    "RNNs process sequences one word at a time",
    "Deep learning models learn from data",
    "Transformers process all words in parallel"
]

# Create embeddings for all sentences
embeddings = model.encode(sentences)

print(f"Created {len(embeddings)} embeddings")
print(f"Each embedding has {embeddings.shape[1]} dimensions")
print(f"Total shape: {embeddings.shape}\n")
```

```python
# Compute cosine similarity between all pairs
similarity_matrix = cosine_similarity(embeddings)

print("=" * 80)
print("SIMILARITY MATRIX (How similar are the sentences to each other?)")
print("=" * 80)
print()

import pandas as pd

# Create a nice table
df = pd.DataFrame(
    similarity_matrix,
    index=[f"Sent {i+1}" for i in range(len(sentences))],
    columns=[f"Sent {i+1}" for i in range(len(sentences))]
)

print(df.round(3))
print()

# Print interpretations
print("=" * 80)
print("INTERPRETATION:")
print("=" * 80)
for i in range(len(sentences)):
    print(f"\nSentence {i+1}: {sentences[i]}")
    print(f"  - Similarity to Sent 1: {similarity_matrix[i, 0]:.3f}")
    print(f"  - Similarity to Sent 2: {similarity_matrix[i, 1]:.3f}")
    print(f"  - Similarity to Sent 3: {similarity_matrix[i, 2]:.3f}")
    print(f"  - Similarity to Sent 4: {similarity_matrix[i, 3]:.3f}")
    print(f"  - Similarity to Sent 5: {similarity_matrix[i, 4]:.3f}")
```

```python
# Documents in our "knowledge base"
documents = [
    "Transformers use self-attention to understand relationships between words",
    "Python is a popular programming language for machine learning",
    "The attention mechanism allows each word to attend to every other word",
    "Deep neural networks with many layers are called deep learning models",
    "BERT (Bidirectional Encoder Representations) is based on transformer architecture"
]

# User query
query = "How do transformers work?"

print("="*80)
print("RAG RETRIEVAL SIMULATION")
print("="*80)
print(f"\nQuery: {query}")
print(f"\nKnowledge base has {len(documents)} documents")
print()

# Encode documents and query
doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])[0]  # Get first (only) embedding

# Compute similarity between query and all documents
similarities = cosine_similarity([query_embedding], doc_embeddings)[0]

# Sort by similarity (highest first)
ranked = sorted(enumerate(similarities), key=lambda x: x[1], reverse=True)

# Display results
print("TOP RESULTS (by relevance):")
print()

for rank, (doc_idx, score) in enumerate(ranked, 1):
    print(f"{rank}. Similarity: {score:.3f} | Document: {documents[doc_idx]}")
    if score >= 0.7:
        print("   → ✓ HIGHLY RELEVANT (>0.7)")
    elif score >= 0.5:
        print("   → ◐ SOMEWHAT RELEVANT (0.5-0.7)")
    else:
        print("   → ✗ NOT RELEVANT (<0.5)")
    print()

# RAG decision
print("="*80)
print("RAG DECISION:")
print("="*80)
print(f"\nWould retrieve top 2-3 documents (similarity > 0.5):")
retrieved_count = sum(1 for _, score in ranked if score >= 0.5)
print(f"  → {retrieved_count} documents qualify")
print(f"\nThese would be passed to LLM along with the query.")
print(f"LLM would answer USING ONLY these documents (grounded answer).")
```

```python
# ============================================
# EDIT THESE SENTENCES AND RUN AGAIN
# ============================================

test_pairs = [
    ("Machine learning is a subset of artificial intelligence",
     "AI and ML are different fields"),

    ("The weather is sunny today",
     "It is raining outside"),

    ("Neural networks learn from training data",
     "Deep learning models are trained using backpropagation"),
]

print("="*80)
print("SIMILARITY ANALYSIS - SENTENCE PAIRS")
print("="*80)
print()

for i, (sent1, sent2) in enumerate(test_pairs, 1):
    emb1 = model.encode([sent1])[0]
    emb2 = model.encode([sent2])[0]
    sim = cosine_similarity([emb1], [emb2])[0][0]

    print(f"Pair {i}:")
    print(f"  Sentence 1: {sent1}")
    print(f"  Sentence 2: {sent2}")
    print(f"  Similarity: {sim:.4f}")

    if sim >= 0.8:
        print(f"  → VERY SIMILAR (same topic, similar language)")
    elif sim >= 0.6:
        print(f"  → SIMILAR (related concepts)")
    elif sim >= 0.4:
        print(f"  → SOMEWHAT SIMILAR (some connection)")
    else:
        print(f"  → DIFFERENT (unrelated topics)")
    print()
```

```python
# Initialize text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=256,        # Characters per chunk (not words)
    chunk_overlap=50,      # Overlap for context continuity
    separators=["\n\n", "\n", ". ", " ", ""]  # How to split (recursive)
)

# Chunk all documents
all_chunks = []
chunk_metadata = []

for doc_name, doc_text in documents.items():
    chunks = text_splitter.split_text(doc_text)

    print(f"\n📄 Document: {doc_name}")
    print(f"   Original length: {len(doc_text)} characters")
    print(f"   Number of chunks: {len(chunks)}")
    print(f"   Avg chunk size: {len(doc_text) // len(chunks)} characters")

    # Store chunks with metadata
    for i, chunk in enumerate(chunks):
        all_chunks.append(chunk)
        chunk_metadata.append({
            "source": doc_name,
            "chunk_id": i,
            "chunk_number": f"{i+1}/{len(chunks)}"
        })

    # Show first chunk as example
    print(f"   First chunk preview:")
    print(f"   '{chunks[0][:100]}...'\n")

print(f"\n✓ Total chunks created: {len(all_chunks)}")
```

```python
# Initialize Chroma client (persistent)
chroma_client = chromadb.PersistentClient(path=chroma_folder)

print("✓ Chroma client initialized (persistent mode)")
print(f"  Storage location: {chroma_folder}")
```

```python
# Delete existing collection if it exists (fresh start)
try:
    chroma_client.delete_collection("rag_collection")
    print("Deleted existing collection")
except:
    pass

# Create new collection
collection = chroma_client.create_collection(
    name="rag_collection",
    metadata={"hnsw:space": "cosine"}  # Use cosine similarity
)

print("✓ Created new Chroma collection: 'rag_collection'")

# Add documents to collection
print(f"\nAdding {len(all_chunks)} chunks to Chroma...")

# Chroma needs: ids, documents, metadatas
ids = [f"chunk_{i}" for i in range(len(all_chunks))]

collection.add(
    ids=ids,
    documents=all_chunks,
    metadatas=chunk_metadata,
    embeddings=None  # Chroma will generate embeddings automatically
)

print(f"✓ Added all chunks to Chroma!")
print(f"  Collection size: {collection.count()} chunks")
```

```python
# Test Query 1: About Transformers
query1 = "How do transformers use attention?"

print("="*80)
print(f"Query: {query1}")
print("="*80)

results1 = collection.query(
    query_texts=[query1],
    n_results=3  # Get top 3 most relevant chunks
)

print(f"\nTop 3 Results:")
print()
for i, (chunk, metadata, distance) in enumerate(
    zip(results1['documents'][0], results1['metadatas'][0], results1['distances'][0]),
    1
):
    # Convert distance to similarity (Chroma returns distance, we want similarity)
    similarity = 1 - distance

    print(f"{i}. Relevance Score: {similarity:.3f}")
    print(f"   Source: {metadata['source']} (chunk {metadata['chunk_number']})")
    print(f"   Text: {chunk[:150]}...")
    print()

print("✓ Retrieval working!")
```

```python
# Test multiple queries from different topics
test_queries = [
    "What is self-attention mechanism?",
    "Tell me about SQL databases",
    "How does backpropagation work?",
    "Explain Python decorators",
    "What is BERT?",
    "NoSQL vs SQL databases"
]

# Function to display results nicely
def display_retrieval_results(query, results, top_k=2):
    print(f"\n{'='*80}")
    print(f"Query: {query}")
    print(f"{'='*80}")

    for rank, (chunk, metadata, distance) in enumerate(
        zip(results['documents'][0][:top_k],
            results['metadatas'][0][:top_k],
            results['distances'][0][:top_k]),
        1
    ):
        similarity = 1 - distance

        # Relevance rating
        if similarity >= 0.7:
            rating = "✓✓✓ HIGHLY RELEVANT"
        elif similarity >= 0.5:
            rating = "✓✓ RELEVANT"
        elif similarity >= 0.3:
            rating = "✓ SOMEWHAT RELEVANT"
        else:
            rating = "✗ NOT RELEVANT"

        print(f"\n{rank}. [{rating}] Score: {similarity:.3f}")
        print(f"   From: {metadata['source']} ({metadata['chunk_number']})")
        print(f"   \"{chunk[:120]}...\"")

# Run all test queries
for query in test_queries:
    results = collection.query(
        query_texts=[query],
        n_results=3
    )
    display_retrieval_results(query, results, top_k=2)
```

```python
# ============================================
# EDIT THESE QUERIES AND RUN
# ============================================

custom_queries = [
    "What are the advantages of transformers?",
    "How to normalize databases?",
    "What is overfitting?",
]

print("\n" + "="*80)
print("CUSTOM QUERY TESTING")
print("="*80)

for query in custom_queries:
    results = collection.query(
        query_texts=[query],
        n_results=3
    )
    display_retrieval_results(query, results, top_k=3)
```

```python
# Comprehensive evaluation
print("="*80)
print("RETRIEVAL QUALITY ANALYSIS")
print("="*80)

evaluation_queries = {
    "transformers": ["What is a transformer?", "How do transformers work?"],
    "databases": ["What is a database?", "SQL vs NoSQL"],
    "deep_learning": ["What is backpropagation?", "Activation functions"],
    "python": ["Python best practices", "Object oriented programming"]
}

for category, queries in evaluation_queries.items():
    print(f"\n{'='*80}")
    print(f"Category: {category.upper()}")
    print(f"{'='*80}")

    for query in queries:
        results = collection.query(
            query_texts=[query],
            n_results=5
        )

        # Get top scores
        top_similarity = 1 - results['distances'][0][0]
        retrieved_source = results['metadatas'][0][0]['source']

        # Check if correct source
        correct = "✓" if category in retrieved_source else "✗"

        print(f"\nQuery: {query}")
        print(f"  Top result: {retrieved_source}")
        print(f"  Similarity: {top_similarity:.3f}")
        print(f"  Correct: {correct}")
```
